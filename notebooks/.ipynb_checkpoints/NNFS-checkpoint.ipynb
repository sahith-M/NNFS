```python
#A single neuron with 4 inputs
```

```python
inputs=[1,2,3,2.5]
weights=[0.2,0.8,-0.5,1.0]
bias=2
```

```python
output=(inputs[0]*weights[0]+inputs[1]*weights[1]+inputs[2]*weights[2] + inputs[3]*weights[3]+bias)
print(output)
```

```python
#Implementing a layer of neurons
```

```python
inputs=[1,2,3,2.5]
weights1=[0.2,0.8,-0.5,1.0]
weights2 = [0.5, -0.91, 0.26, -0.5] 
weights3 = [-0.26, -0.27, 0.17, 0.87] 

bias1=2
bias2 = 3 
bias3 = 0.5 
```

```python
outputs = [ 
        # Neuron 1: 
        inputs[0]*weights1[0] + 
        inputs[1]*weights1[1] + 
        inputs[2]*weights1[2] + 
        inputs[3]*weights1[3] + bias1, 
 
        # Neuron 2: 
        inputs[0]*weights2[0] + 
        inputs[1]*weights2[1] + 
        inputs[2]*weights2[2] + 
        inputs[3]*weights2[3] + bias2, 
 
        # Neuron 3: 
        inputs[0]*weights3[0] + 
        inputs[1]*weights3[1] + 
        inputs[2]*weights3[2] + 
        inputs[3]*weights3[3] + bias3] 
 
print(outputs) 
```

```python
#Implementing NN using loop to scale and handle dynamically-sized inputs and layers
```

```python
inputs = [1, 2, 3, 2.5] 
weights = [[0.2, 0.8, -0.5, 1], 
           [0.5, -0.91, 0.26, -0.5], 
           [-0.26, -0.27, 0.17, 0.87]] 
biases = [2, 3, 0.5] 

layer_output=[]
for input_weight,input_bias in zip(weights,biases):
    n_output=0
    for n_input,weight in zip(inputs,input_weight):
        n_output=n_input*weight + n_output
    n_output=n_output+input_bias
    layer_output.append(n_output)
```

```python
#Single Neuron using dot product
```

```python
import numpy as np
inputs = [1, 2, 3, 2.5] 
weights=[0.2,0.8,-0.5,1.0]
bias=2
```

```python
outputs=np.dot(weights,inputs) + bias
print(outputs)
```

```python
#Implementing layer of neurons using dot product
```

```python
inputs = [1, 2, 3, 2.5] 
weights = [[0.2, 0.8, -0.5, 1], 
           [0.5, -0.91, 0.26, -0.5], 
           [-0.26, -0.27, 0.17, 0.87]] 
biases = [2, 3, 0.5]

layer_outputs=np.dot(weights,inputs)+biases
print(layer_outputs)
```

```python
#A Layer of Neurons & Batch of Data w/ NumPy 
```

```python
inputs = [[1, 2, 3, 2.5],
          [2,5,-1,2],
          [-1.5,2.7,3.3,-0.8] ]
weights = [[0.2, 0.8, -0.5, 1], 
           [0.5, -0.91, 0.26, -0.5], 
           [-0.26, -0.27, 0.17, 0.87]] 
biases = [2, 3, 0.5]
outputs=np.dot(inputs,np.array(weights).T)+biases
print(outputs)
```

```python
#Adding layers
```

```python
inputs = [[1, 2, 3, 2.5],
          [2,5,-1,2],
          [-1.5,2.7,3.3,-0.8] ]
weights = [[0.2, 0.8, -0.5, 1], 
           [0.5, -0.91, 0.26, -0.5], 
           [-0.26, -0.27, 0.17, 0.87]] 
biases = [2, 3, 0.5]
weights2 = [[0.1, -0.14, 0.5], 
           [-0.5, 0.12, -0.33], 
           [-0.44, 0.73, -0.13]] 
biases2 = [-1, 2, -0.5]
layer1_outputs=np.dot(inputs,np.array(weights).T) + biases
layer2_outputs=np.dot(layer1_outputs,np.array(weights2).T) + biases2
print(layer2_outputs)
```

```python
#Dense Layer Class
```

```python
class Layer_Dense:
    def __init__(self,n_inputs,n_neurons):
        self.weights=0.01*np.random.randn(n_inputs,n_neurons)
        self.biases= np.zeros((1,n_neurons))

    def forward(self, inputs):
        self.output=np.dot(inputs, self.weights) + self.biases
X,y=spiral_data(samples=100, classes=3)

dense1=Layer_Dense(2, 3)
dense1.forward(X)

print(dense1.output[:5])
```
